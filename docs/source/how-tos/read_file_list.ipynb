# How to read multiple files as a plain list, without merging?

By default, when *earthkit-data* reads multiple files it tries to automatically merge them into a
single object (e.g. a single fieldlist when they are all GRIB). Sometimes, however, we are not
interested in the merged data itself, but simply want the list of the input files that were found.

## Reading a tar archive of unsupported files

```python
import earthkit.data as ekd

ds = ekd.from_source(
    "url",
    "https://sites.ecmwf.int/repository/earthkit-data/test-data/binary_unknown.tar",
)
```

```python
ds.describe()
```

```python
ds = ekd.from_source(
    "url",
    "https://sites.ecmwf.int/repository/earthkit-data/test-data/binary_unknown.tar",
    merger=False,
)
```

```python
ds.describe()
```

## Getting the list of input files

The ``path`` property of the returned object gives us the list of the individual input files:

```python
ds.path
```

We can also iterate through the data items:

```python
for d in ds:
    print(d)
```
