# How to read multiple files as a plain list, without merging?

By default, when *earthkit-data* reads multiple files it tries to automatically merge them into a
single object (e.g. a single fieldlist when they are all GRIB). Sometimes, however, we are not
interested in the merged data itself, but simply want the list of the input files that were found
-- e.g. to pass them on to another tool, or to inspect what *earthkit-data* actually resolved the
input to.

## Reading a tar archive of GRIB files

```python
import earthkit.data as ekd

ds = ekd.from_source(
    "url",
    "https://sites.ecmwf.int/repository/earthkit-data/examples/test_gribs.tar.gz",
    merger=False,
)
```

```python
type(ds).__name__
```

## Getting the list of input files

The ``path`` property of the returned object gives us the list of the individual input files:

```python
ds.path
```

We can also check how many files were found:

```python
len(ds.path)
```

If we still need the actual data, we can convert ``ds`` into a fieldlist, exactly as we would for
the default, merging behaviour:

```python
fl = ds.to_fieldlist()
len(fl)
```
