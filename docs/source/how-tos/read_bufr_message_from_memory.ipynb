# How to read a single BUFR message from memory?

A single BUFR message can be read from memory the same way as GRIB. However, since BUFR data is not organised into Fields, the result is a FeatureList rather than a FieldList.

```python
import earthkit.data as ekd

ds = ekd.from_source("sample", "temp_10.bufr").to_featurelist()
buf = ds[0].message()
```

```python
bf = ekd.from_source("memory", buf).to_featurelist()[0]
bf
```
