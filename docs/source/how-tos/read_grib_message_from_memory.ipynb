# How to read a single GRIB message from memory?

Sometimes we already have the raw bytes of a single GRIB message in memory (e.g. received over the network) and want to turn it into a Field without first writing it to a file.

```python
import earthkit.data as ekd

ds = ekd.from_source("sample", "storm_ophelia_wind_850.grib").to_fieldlist()
buf = ds[0].message()
```

```python
field = ekd.from_source("memory", buf).to_fieldlist()[0]
field
```
