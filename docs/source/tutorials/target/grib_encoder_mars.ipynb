# Encoding GRIB from MARS metadata and values

```python
import numpy as np

import earthkit.data as ekd
```

#### Encoding a surface field on a regular lat-lon grid

```python
shape = (181, 360)
values = np.random.random(shape) * 100.0 + 200  # temperature like values

mars_metadata = {
    "origin": "ecmf",
    "type": "fc",
    "class": "od",
    "stream": "oper",
    "expver": "0001",
    "date": 20010101,
    "time": 0,
    "step": 6,
    "param": 167,
    "levtype": "sfc",
    "packing": "ccsds",
    "grid": "1/1",
}

encoder = ekd.create_encoder("grib")
r = encoder.encode(mars_metadata=mars_metadata, values=values)
f = r.to_field()
f
```

#### Using `mars_metadata` together with `metadata`

```python
shape = (40320,)
values = np.random.random(shape) * 100.0 + 200  # temperature like values

mars_metadata = {
    "origin": "ecmf",
    "type": "fc",
    "class": "od",
    "stream": "oper",
    "expver": "0001",
    "date": 20010101,
    "time": 0,
    "step": 6,
    "param": 167,
    "levtype": "sfc",
    "packing": "ccsds",
    "grid": "o96",
}

metadata = {"bitsPerValue": 12}

r = encoder.encode(mars_metadata=mars_metadata, metadata=metadata, values=values)
f = r.to_field()
```
