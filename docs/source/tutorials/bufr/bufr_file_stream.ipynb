# BUFR: reading data as a stream

```python
import earthkit.data as ekd

ekd.download_example_file("temp_10.bufr")
```

## Reading messages one at a time

```python
ds_in = ekd.from_source("file", "temp_10.bufr", stream=True)
ds_in
```

Convert the data source to a featurelist to iterate over its BUFR messages. A stream featurelist is consumed as it is iterated, so its length is not available before reading.

```python
ds = ds_in.to_featurelist()

for message in ds:
    print(message)
```

After iteration, the stream has been consumed and no messages remain to read.

```python
sum(1 for _ in ds)
```

## Reading in batches

Use `batched()` to process a fixed number of BUFR messages at a time. The final batch can contain fewer messages than the requested batch size.

```python
ds = ekd.from_source("file", "temp_10.bufr", stream=True).to_featurelist()

for batch in ds.batched(3):
    print(f"len={len(batch)} dataCategory={batch.metadata('dataCategory')}")
```

## Loading all messages into memory

Pass `read_all=True` to `to_featurelist()` to load the complete stream into memory. The resulting featurelist supports operations such as inspecting its length and metadata.

```python
ds = ekd.from_source("file", "temp_10.bufr", stream=True).to_featurelist(read_all=True)

len(ds)
```
