# Using SQL data

```python
db_path = "obs.sqlite"
```

## Reading an SQLite file with the `file` source

```python
import earthkit.data as ekd

ds = ekd.from_source("file", db_path)
ds
```

```python
try:
    ds.to_pandas()
except ValueError as e:
    print(e)
```

```python
ds.to_pandas(table="obs")
```

```python
ds.to_pandas(table="stations")
```

## Reading geometries with `to_geopandas()`

## Running a query with the `sql` source

```python
ds = ekd.from_source("sql", f"sqlite:///{db_path}", "SELECT station, t2m, valid FROM obs WHERE t2m > 285")
ds
```

```python
ds.to_pandas()
```

```python
ds = ekd.from_source("sql", f"sqlite:///{db_path}", "SELECT station, t2m FROM obs WHERE t2m > :t")
ds.to_pandas(params={"t": 285})
```

```python
gdf = ekd.from_source("sql", f"sqlite:///{db_path}", "SELECT * FROM obs").to_geopandas(crs="EPSG:4326")
gdf
```
