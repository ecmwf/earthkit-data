# EWDS: retrieving data

```python
import earthkit.data as ekd
```

### Getting GRIB data

```python
d = ekd.from_source(
    "ewds",
    "cems-fire-seasonal",
    request=dict(
        variable="fire_weather_index",
        release_version="5",
        year="1991",
        month="09",
        day="01",
        leadtime_hour="12",
        data_format="grib",
    ),
)
```

The request contains 25 fields:

```python
fl = d.to_fieldlist()
len(fl)
```

```python
fl.head()
```

### Getting NetCDF data

The output format is specified with the ``data_format`` request key (not ``format``):

```python
d1 = ekd.from_source(
    "ewds",
    "cems-fire-seasonal",
    request=dict(
        variable="fire_weather_index",
        release_version="5",
        year="1991",
        month="09",
        day="01",
        leadtime_hour="12",
        data_format="netcdf",
    ),
)
```

```python
ds = d1.to_xarray()
ds.attrs.pop("history", None)  # contains internal temporary file paths
ds
```

### Splitting a request

```python
d2 = ekd.from_source(
    "ewds",
    "cems-fire-seasonal",
    request=dict(
        variable="fire_weather_index",
        release_version="5",
        year="1991",
        month="09",
        day="01",
        leadtime_hour=["12", "36"],
        data_format="grib",
        split_on="leadtime_hour",
    ),
)
len(d2.to_fieldlist())
```
