# Zenodo: accessing files inside a record

```python
import earthkit.data as ekd

d = ekd.from_source(
    "zenodo", identifier="10.5281/zenodo.13918054", filter=["cat_ens_20240520_pf.grb", "cat_ens_20240521_pf.grb"]
)
d
```

```python
d.to_fieldlist().head()
```

The `identifier` accepts both DOIs and Zenodo record numbers and resolves concept IDs to the latest version of the corresponding record.

The `filter` argument supports explicit selection of files by listing them as in the example above. File names can also be matched with a shell-style wildcard pattern:

```python
ekd.from_source("zenodo", identifier=13918054, filter="*_20240520_*")
```

For more control, `filter` can also be a callable. It is called with each file name in the record and should return `True` for the files to select:

```python
ekd.from_source("zenodo", identifier=13918054, filter=lambda name: "20240521" in name and "noconv" in name)
```

If no `filter` is provided, all files of the record are retrieved.
