# CMEMS: retrieving data

```python
import earthkit.data as ekd
```

### Retrieving NetCDF data

```python
request = dict(
    variables=["thetao"],
    minimum_longitude=-10,
    maximum_longitude=-9,
    minimum_latitude=40,
    maximum_latitude=41,
    start_datetime="2020-01-01",
    end_datetime="2020-01-01T23:59:59",
    minimum_depth=0,
    maximum_depth=1,
)

d = ekd.from_source("cmems", "cmems_mod_ibi_phy-temp_my_0.027deg_P1M-m", request=request)
d
```

```python
ds = d.to_xarray()
ds
```

### Retrieving CSV data

```python
d_csv = ekd.from_source(
    "cmems",
    "cmems_mod_ibi_phy-temp_my_0.027deg_P1M-m",
    request=dict(request, file_format="csv"),
)
d_csv
```

```python
df = d_csv.to_pandas()
df.head()
```
